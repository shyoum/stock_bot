```python
import FinanceDataReader as fdr
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

nasdaq_tickers = fdr.StockListing('NASDAQ')['Symbol'].tolist()
np.random.shuffle(nasdaq_tickers)

# 100개 뽑고 필터링하여 유효한 것만 남기고 부족하면 추가
recent_date = fdr.DataReader('AAPL', '2025-01-01').index[-1].strftime('%Y-%m-%d')
valid_tickers = []
idx = 0

while len(valid_tickers) < 100 and idx < len(nasdaq_tickers):
    t = nasdaq_tickers[idx]
    try:
        if not fdr.DataReader(t, '2025-01-01', '2025-01-01').empty and not fdr.DataReader(t, recent_date, recent_date).empty:
            valid_tickers.append(t)
    except:
        pass
    idx += 1

nasdaq_tickers = valid_tickers

for i in range(0, min(100, len(nasdaq_tickers)), 10):
    globals()[f"nt_{i+10}"] = nasdaq_tickers[i:i+10]

# 첫 10개 종목 출력 및 데이터 확인
if len(nasdaq_tickers) >= 10:
    for i in range(10):
        print(nasdaq_tickers[i])
        df = fdr.DataReader(nasdaq_tickers[i], '2025-08-25')[['Open', 'Close', 'Volume']]
        display(df)
```
