```python
import FinanceDataReader as fdr
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

nasdaq_tickers = fdr.StockListing('NASDAQ')['Symbol'].tolist()
np.random.shuffle(nasdaq_tickers)

# 방법 1: for 반복문으로 10개씩 출력
for i in range(0, 100, 10):
    start_idx = i
    end_idx = i + 10
    nt = nasdaq_tickers[start_idx:end_idx]
    print(f"Group {i//10 + 1}: {nt}")

# 첫 번째 그룹 따로 저장
first_group = nasdaq_tickers[:10]
print(f"\n=== 첫 번째 그룹 주식 데이터 조회 ===")
print(f"대상 종목: {first_group}")

# 최근 거래일 설정 (오늘부터 최대 5일 전까지 확인)
end_date = datetime.now().strftime('%Y-%m-%d')
start_date = (datetime.now() - timedelta(days=5)).strftime('%Y-%m-%d')

for idx, ticker in enumerate(first_group):
    try:
        print(f"\n[{idx+1}] {ticker} 주식 데이터:")
        print("-" * 40)
        
        # 주식 데이터 조회
        stock_data = fdr.DataReader(ticker, start_date, end_date)
        
        if not stock_data.empty:
            # 최근 거래일 데이터 (마지막 행)
            latest_data = stock_data.iloc[-1:]
            print(latest_data)
            
            # 간단한 요약 정보
            latest_close = stock_data['Close'].iloc[-1]
            latest_volume = stock_data['Volume'].iloc[-1]
            print(f"최근 종가: ${latest_close:.2f}")
            print(f"최근 거래량: {latest_volume:,}")
            
        else:
            print(f"{ticker}: 데이터를 찾을 수 없습니다.")
            
    except Exception as e:
        print(f"{ticker}: 오류 발생 - {str(e)}")
```
