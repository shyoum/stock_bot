```python
import os
import glob

def update_menu():
    """현재 폴더의 모든 .py 파일을 menu.py에 자동 등록"""
    
    # 현재 폴더의 모든 .py 파일 찾기 (menu.py와 자기 자신 제외)
    py_files = []
    for file in glob.glob("*.py"):
        if file not in ["menu.py", "update_menu.py", "__init__.py"]:
            py_files.append(file)
    
    if not py_files:
        print("등록할 .py 파일이 없습니다.")
        return
    
    # menu.py 생성
    menu_content = generate_menu_content(py_files)
    
    with open("menu.py", "w", encoding="utf-8") as f:
        f.write(menu_content)
    
    print(f"✅ menu.py 업데이트 완료!")
    print(f"📁 등록된 파일들:")
    for file in py_files:
        page_name = get_page_name(file)
        print(f"   • {page_name} ({file})")

def get_page_name(file_name):
    """파일명을 페이지명으로 변환"""
    name = file_name.replace(".py", "").replace("_", " ")
    return name.title()

def get_module_name(file_name):
    """파일명을 모듈명으로 변환"""
    return file_name.replace(".py", "")

def generate_menu_content(py_files):
    """menu.py 내용 생성"""
    
    # Import 섹션 생성
    imports = ["import streamlit as st\n"]
    for file in py_files:
        module = get_module_name(file)
        imports.append(f"from {module} import main as {module}_main")
    
    # Pages 딕셔너리 생성
    pages = []
    for file in py_files:
        page_name = get_page_name(file)
        module = get_module_name(file)
        pages.append(f'    "{page_name}": {module}_main,')
    
    # 전체 내용 조합
    content = f'''{chr(10).join(imports)}

# 자동 생성된 페이지 메뉴
pages = {{
{chr(10).join(pages)}
}}

def show_menu():
    """사이드바에 메뉴 표시"""
    st.sidebar.title("📋 메뉴")
    
    if pages:
        selected_page = st.sidebar.selectbox(
            "페이지 선택",
            list(pages.keys())
        )
        
        if selected_page:
            pages[selected_page]()
    else:
        st.sidebar.write("등록된 페이지가 없습니다.")

if __name__ == "__main__":
    show_menu()
'''
    
    return content

if __name__ == "__main__":
    update_menu()
```
